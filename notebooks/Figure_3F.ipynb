```python
fig, ax = plt.subplots(figsize=(2.3, 3.4), dpi=150)
plt.rcParams["font.family"] = "Arial"

bp = ax.boxplot(
    [normal_counts, aml_counts],
    patch_artist=True,
    widths=0.5,
    whis=1.5,
    showfliers=True
)

colors = ['#4C90C0', '#F2545B']

for patch, color in zip(bp['boxes'], colors):
    patch.set_facecolor(color)
    patch.set_edgecolor(color)
    patch.set_linewidth(1.1)

for whisker, color in zip(bp['whiskers'], [colors[0], colors[0], colors[1], colors[1]]):
    whisker.set_color(color)
    whisker.set_linewidth(1.0)

for cap, color in zip(bp['caps'], [colors[0], colors[0], colors[1], colors[1]]):
    cap.set_color(color)
    cap.set_linewidth(1.0)

for median in bp['medians']:
    median.set_visible(False)

for flier, color in zip(bp['fliers'], colors):
    flier.set_marker('o')
    flier.set_markerfacecolor(color)
    flier.set_markeredgecolor(color)
    flier.set_markersize(1.8)
    flier.set_alpha(1)

ax.set_xticks([1, 2])
ax.set_xticklabels(['Normal', 'AML'], fontsize=14)
ax.set_ylabel('ALLocate Blast Count', fontsize=14, fontweight='bold')

ax.set_ylim(-0.5, 50)
ax.tick_params(axis='y', labelsize=10, length=0)
ax.tick_params(axis='x', length=0)

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_linewidth(0.8)
ax.spines['bottom'].set_linewidth(0.8)


plt.tight_layout()
plt.show()
```
